```python
# Install first (run once):
# pip install qrcode[pil]

from pathlib import Path
import qrcode
from qrcode.constants import ERROR_CORRECT_M

OUTPUTS_DIR = Path("outputs")
OUTPUTS_DIR.mkdir(parents=True, exist_ok=True)

def generate_qr_png(link: str, filename: str = "qrcode.png",
                    box_size: int = 10, border: int = 4,
                    error_correction=ERROR_CORRECT_M,
                    output_dir: Path = OUTPUTS_DIR):
    """
    Generate a PNG QR code for `link` and save to `output_dir / filename`.
    - box_size: pixel size of each box
    - border: boxes of border (minimum 4 recommended)
    - error_correction: one of qrcode.constants.ERROR_CORRECT_{L,M,Q,H}
    - output_dir: directory to save the output QR code
    """
    qr = qrcode.QRCode(
        version=None,  # automatic size
        error_correction=error_correction,
        box_size=box_size,
        border=border,
    )
    qr.add_data(link)
    qr.make(fit=True)

    img = qr.make_image(fill_color="black", back_color="white")
    out_path = Path(filename)
    if not out_path.is_absolute():
        out_path = output_dir / out_path
    img.save(out_path)
    print(f"Saved PNG QR code to: {out_path}")

# Example usage:
if __name__ == "__main__":
    generate_qr_png("https://youtu.be/P79jQq110c4", "ជំនួយការ AI ពហុមុខងារ 0.02-beta — របៀបដំឡើង និងប្រើប្រាស់ (LocalAiLab).png")
```
